## 1. Upload a swing video
One golfer, full swing, phone held still.

```python
from google.colab import files
up = files.upload()
video = next(iter(up))
print(video)
```

## 2. Trace it
Modes: `both` (boxes + path), `path`, `boxes`, `compare`.

```python
import swingtrace, torch
device = 0 if torch.cuda.is_available() else 'cpu'
for mode in ['both', 'compare']:
    r = swingtrace.track(video, mode=mode, device=device)
    print(mode, '->', r.summary())
```

## 3. Watch the result

```python
import subprocess, base64
from pathlib import Path
from IPython.display import HTML
src = f"out/{Path(video).stem}_both.mp4"
subprocess.run(['ffmpeg','-v','error','-y','-i',src,'-c:v','libx264','-pix_fmt','yuv420p','web.mp4'], check=True)
HTML(f'<video width=360 controls src="data:video/mp4;base64,{base64.b64encode(open("web.mp4","rb").read()).decode()}"></video>')
```

## 4. Download


```python
files.download(f"out/{Path(video).stem}_both.mp4")
```
